```python

# Function to fit kinetics for each unique condition
def fit_kinetics_per_condition(df):
    """
    Fit exponential decay for each unique combination of Initial Concentration and Temperature
    """
    results = []

    for (A0, T), group in df.groupby(['Initial Concentration', 'Temperature']):

        # Define model with this group's specific A0
        model = lambda t, k: A0 * np.exp(-k * t)

        # Fit rate constant
        k_fit, _ = curve_fit(model, group['Time'], group['A'], p0=[0.5])
        k_fit = k_fit[0]

        # Store results
        results.append({
            'A0': A0,
            'Temperature': T,
            'k': k_fit,
            'equation': f'A(t) = {A0} * exp(-{k_fit:.4f} * t)',
            'time_data': group['Time'].values,
            'conc_data': group['A'].values
        })

    return results

# Apply to your data
kinetic_results = fit_kinetics_per_condition(df1)

# Print all equations
for result in kinetic_results:
    print(f"Initial A₀: {result['A0']}, Temperature: {result['Temperature']}K")
    print(f"Fitted equation: {result['equation']}")
    print()
```

```python
def consecutive_model_B(t, k1, k2, A0):
    """
    Kinetic model for an intermediate B in A -> B -> I.
    A0 is passed as a fixed argument.
    """
    if abs(k1 - k2) < 1e-9:
        return A0 * k1 * t * np.exp(-k1 * t)
    else:
        return A0 * (k1 / (k2 - k1)) * (np.exp(-k1 * t) - np.exp(-k2 * t))

# --- 3. Group data and fit each experimental run ---
fit_results = []
plt.figure(figsize=(25, 15))

# Group by each unique combination of conditions
for (A0, temp), group in df2.groupby(['Initial', 'Temperature']):

    t_data = group['Time'].values
    B_data = group['B'].values

    # --- Fit k1 and k2 for this specific run, using its A0 ---
    # We use a lambda function to pass A0 as a fixed parameter to the model.
    try:
        params, _ = curve_fit(
            lambda t, k1, k2: consecutive_model_B(t, k1, k2, A0=A0),
            t_data,
            B_data,
            p0=[0.5, 0.1] # Initial guesses for k1, k2
        )
        k1_fit, k2_fit = params

        # --- Store the results for this run ---
        fit_results.append({
            'Initial Concentration': A0,
            'Temperature': temp,
            'k1': k1_fit,
            'k2': k2_fit
        })

        # --- Plot the data points and the fitted curve for this run ---
        t_smooth = np.linspace(t_data.min(), t_data.max(), 200)
        B_fit = consecutive_model_B(t_smooth, k1_fit, k2_fit, A0)

        plot = plt.scatter(t_data, B_data, s=50, label=f'Data (A₀={A0}, T={temp}K)')
        plt.plot(t_smooth, B_fit, color=plot.get_facecolor()[0], label=f'Fit (A₀={A0}, T={temp}K)')

    except RuntimeError:
        print(f"Fit failed for A₀={A0}, T={temp}K. Check data or initial guesses.")

# --- 4. Display results ---
plt.title('Consecutive Reaction Fits for Each Experimental Run')
plt.xlabel('Time')
plt.ylabel('Concentration of B')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

print("\n--- Fitted Rate Constants for Each Run ---")
for result in fit_results:
    print(f"A₀ = {result['Initial Concentration']}, Temp = {result['Temperature']}K  =>  k1={result['k1']:.4f}, k2={result['k2']:.4f}")
```

```python
from scipy.integrate import odeint
def kinetic_model_odes(y, t, k1, k2):
    """
    Defines the system of Ordinary Differential Equations (ODEs).
    y: array of concentrations [A, B, I]
    t: time
    k1, k2: rate constants
    """
    A, B, I = y
    dAdt = -k1 * A
    dBdt =  k1 * A - k2 * B
    dIdt =  k2 * B
    return [dAdt, dBdt, dIdt]

# --- 3. Define the global model function for curve_fit ---
def global_fit_model(t_data_combined, k1, k2, A0):
    """
    Solves the ODEs and returns a stacked array of [A, B, I] concentrations.
    This function is what curve_fit will use.
    """
    # Reshape the combined time data back to its original form
    t_single = np.unique(t_data_combined)

    # Set initial conditions for the ODE solver: [A0, B0, I0]
    y0 = [A0, 0, 0]

    # Solve the ODEs
    concentrations = odeint(kinetic_model_odes, y0, t_single, args=(k1, k2))

    # Return the concentrations stacked in the same way as the experimental data
    return concentrations.T.flatten()


# --- 4. Group data and perform a global fit for each run ---
fit_results = []
plt.figure(figsize=(20, 12))

for (A0, temp), group in df2.groupby(['Initial', 'Temperature']):

    t_data = group['Time'].values
    A_data = group['A'].values
    B_data = group['B'].values
    I_data = group['I'].values

    # --- Stack the experimental data: [A_vals, B_vals, I_vals] ---
    y_data_stacked = np.concatenate([A_data, B_data, I_data])
    t_data_stacked = np.concatenate([t_data, t_data, t_data])

    try:
        # --- Perform the global fit ---
        # Use bounds to ensure rate constants are non-negative
        params, _ = curve_fit(
            lambda t, k1, k2: global_fit_model(t, k1, k2, A0=A0),
            t_data_stacked,
            y_data_stacked,
            p0=[0.5, 0.2],         # Better initial guesses help
            bounds=(0, np.inf)    # Force k1, k2 to be positive
        )
        k1_fit, k2_fit = params

        fit_results.append({'A0': A0, 'Temp': temp, 'k1': k1_fit, 'k2': k2_fit})

        # --- Plot the results ---
        t_smooth = np.linspace(t_data.min(), t_data.max(), 200)
        y_fit_smooth = odeint(kinetic_model_odes, [A0, 0, 0], t_smooth, args=(k1_fit, k2_fit))

        #plt.plot(t_smooth, y_fit_smooth[:, 0], color='blue', alpha=0.7) # A fit
        #plt.plot(t_smooth, y_fit_smooth[:, 1], color='orange', alpha=0.7) # B fit
        plt.plot(t_smooth, y_fit_smooth[:, 2], alpha=0.7) # I fit

    except RuntimeError as e:
        print(f"Fit failed for A₀={A0}, T={temp}K: {e}")

# Add scatter plot of original data over the fit lines
#plt.scatter(df2['Time'], df2['A'], c='blue', label='A Data', s=40, edgecolor='k', alpha=0.8)
#plt.scatter(df2['Time'], df2['B'], c='orange', label='B Data', s=40, edgecolor='k', alpha=0.8)
plt.scatter(df2['Time'], df2['I'], color=plot.get_facecolor(), label='I Data', s=40, edgecolor='k', alpha=0.8)

plt.title('Global Fit of Consecutive Reaction Kinetics (A → B → I)')
plt.xlabel('Time')
plt.ylabel('Concentration')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

print("\n--- Fitted Rate Constants (Global Fit) ---")
for res in fit_results:
    print(f"A₀={res['A0']}, T={res['Temp']}K => k1={res['k1']:.4f}, k2={res['k2']:.4f}")
```

```python
def consecutive_model_B(t, k1, k2, A0):
    """
    Kinetic model for an intermediate B in A -> B -> I.
    A0 is passed as a fixed argument.
    """
    if abs(k1 - k2) < 1e-9:
        return A0 * k1 * t * np.exp(-k1 * t)
    else:
        return A0 * (k1 / (k2 - k1)) * (np.exp(-k1 * t) - np.exp(-k2 * t))

# --- 3. Group data and fit each experimental run ---
fit_results = []
plt.figure(figsize=(25, 15))

# Group by each unique combination of conditions
for (A0, temp), group in df3.groupby(['Initial', 'Temperature']):

    t_data = group['Time'].values
    B_data = group['B'].values

    # --- Fit k1 and k2 for this specific run, using its A0 ---
    # We use a lambda function to pass A0 as a fixed parameter to the model.
    try:
        params, _ = curve_fit(
            lambda t, k1, k2: consecutive_model_B(t, k1, k2, A0=A0),
            t_data,
            B_data,
            p0=[0.5, 0.1] # Initial guesses for k1, k2
        )
        k1_fit, k2_fit = params

        # --- Store the results for this run ---
        fit_results.append({
            'Initial Concentration': A0,
            'Temperature': temp,
            'k1': k1_fit,
            'k2': k2_fit
        })

        # --- Plot the data points and the fitted curve for this run ---
        t_smooth = np.linspace(t_data.min(), t_data.max(), 200)
        B_fit = consecutive_model_B(t_smooth, k1_fit, k2_fit, A0)

        plot = plt.scatter(t_data, B_data, s=50, label=f'Data (A₀={A0}, T={temp}K)')
        plt.plot(t_smooth, B_fit, color=plot.get_facecolor()[0], label=f'Fit (A₀={A0}, T={temp}K)')

    except RuntimeError:
        print(f"Fit failed for A₀={A0}, T={temp}K. Check data or initial guesses.")

# --- 4. Display results ---
plt.title('Consecutive Reaction Fits for Each Experimental Run')
plt.xlabel('Time')
plt.ylabel('Concentration of B')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

print("\n--- Fitted Rate Constants for Each Run ---")
for result in fit_results:
    print(f"A₀ = {result['Initial Concentration']}, Temp = {result['Temperature']}K  =>  k1={result['k1']:.4f}, k2={result['k2']:.4f}")
```

```python
def kinetic_model_odes(y, t, k1, k2):
    """
    Defines the system of Ordinary Differential Equations (ODEs).
    y: array of concentrations [A, B, I]
    t: time
    k1, k2: rate constants
    """
    A, B, I = y
    dAdt = -k1 * A
    dBdt =  k1 * A - k2 * B
    dIdt =  k2 * B
    return [dAdt, dBdt, dIdt]

# --- 3. Define the global model function for curve_fit ---
def global_fit_model(t_data_combined, k1, k2, A0):
    """
    Solves the ODEs and returns a stacked array of [A, B, I] concentrations.
    This function is what curve_fit will use.
    """
    # Reshape the combined time data back to its original form
    t_single = np.unique(t_data_combined)

    # Set initial conditions for the ODE solver: [A0, B0, I0]
    y0 = [A0, 0, 0]

    # Solve the ODEs
    concentrations = odeint(kinetic_model_odes, y0, t_single, args=(k1, k2))

    # Return the concentrations stacked in the same way as the experimental data
    return concentrations.T.flatten()


# --- 4. Group data and perform a global fit for each run ---
fit_results = []
plt.figure(figsize=(20, 12))

for (A0, temp), group in df3.groupby(['Initial', 'Temperature']):

    t_data = group['Time'].values
    A_data = group['A'].values
    B_data = group['B'].values
    I_data = group['I'].values

    # --- Stack the experimental data: [A_vals, B_vals, I_vals] ---
    y_data_stacked = np.concatenate([A_data, B_data, I_data])
    t_data_stacked = np.concatenate([t_data, t_data, t_data])

    try:
        # --- Perform the global fit ---
        # Use bounds to ensure rate constants are non-negative
        params, _ = curve_fit(
            lambda t, k1, k2: global_fit_model(t, k1, k2, A0=A0),
            t_data_stacked,
            y_data_stacked,
            p0=[0.5, 0.2],         # Better initial guesses help
            bounds=(0, np.inf)    # Force k1, k2 to be positive
        )
        k1_fit, k2_fit = params

        fit_results.append({'A0': A0, 'Temp': temp, 'k1': k1_fit, 'k2': k2_fit})

        # --- Plot the results ---
        t_smooth = np.linspace(t_data.min(), t_data.max(), 200)
        y_fit_smooth = odeint(kinetic_model_odes, [A0, 0, 0], t_smooth, args=(k1_fit, k2_fit))

        #plt.plot(t_smooth, y_fit_smooth[:, 0], color='blue', alpha=0.7) # A fit
        #plt.plot(t_smooth, y_fit_smooth[:, 1], color='orange', alpha=0.7) # B fit
        plt.plot(t_smooth, y_fit_smooth[:, 2], alpha=0.7) # I fit

    except RuntimeError as e:
        print(f"Fit failed for A₀={A0}, T={temp}K: {e}")

# Add scatter plot of original data over the fit lines
#plt.scatter(df2['Time'], df2['A'], c='blue', label='A Data', s=40, edgecolor='k', alpha=0.8)
#plt.scatter(df2['Time'], df2['B'], c='orange', label='B Data', s=40, edgecolor='k', alpha=0.8)
plt.scatter(df3['Time'], df3['I'], color=plot.get_facecolor(), label='I Data', s=40, edgecolor='k', alpha=0.8)

plt.title('Global Fit of Consecutive Reaction Kinetics (A → B → I)')
plt.xlabel('Time')
plt.ylabel('Concentration')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

print("\n--- Fitted Rate Constants (Global Fit) ---")
for res in fit_results:
    print(f"A₀={res['A0']}, T={res['Temp']}K => k1={res['k1']:.4f}, k2={res['k2']:.4f}")
```
